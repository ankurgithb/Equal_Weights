```python
import pandas as pd 
import numpy as np 
import yfinance as yf 
import math
```

```python
tickers = pd.read_csv("top_50_indian_stocks.csv")
# tickers.head()
```

```python
def fetchMarketCap(tickerList):
    data = yf.download(tickerList, period = '1d', group_by='ticker', auto_adjust=False)

    stocksData = []

    for ticker in tickerList:
        latestPrice = data[ticker]['Close'].iloc[-1]
        marketCap = yf.Ticker(ticker).info.get("marketCap", None)
        stocksData.append({
            "Ticker" : ticker,
            "Market Cap" : marketCap,
            "Latest Price" : latestPrice
        })

    stocksDF = pd.DataFrame(stocksData)
    return stocksDF

tickerList = tickers["Ticker"].values.tolist()
df = fetchMarketCap(tickerList)
df.head()
```

```python
df = df.sort_values(by = "Market Cap", ascending = False)
df 
```

```python
df = df.head(10)
df.reset_index(inplace=True, drop = True)
df
```

```python
portfolioSize = int(input("Enter the capital you want to invest: ₹"))
positionSize = portfolioSize / len(df.index)
positionSize
```

```python
df['Number of shares to buy'] = df['Latest Price'].apply(lambda price: math.floor(
    positionSize / price
))
df
```
